````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# 1. Чистим города: strip() убирает пробелы по краям, title() делает "Almaty" из "almaty"/"ALMATY"
cities_clean = []
for c in cities:
    cities_clean.append(c.strip().title())

print(cities_clean)
````

````python
# 2. Строки -> числа
amounts_int = []
for a in amounts:
    amounts_int.append(int(a))

print(amounts_int)
````

````python
# 3. Число заказов, общая сумма, средний чек
orders_count = len(amounts_int)
total_amount = sum(amounts_int)
avg_check = round(total_amount / orders_count)

print("Заказов:", orders_count)
print("Общая сумма:", total_amount, "₸")
print("Средний чек:", avg_check, "₸")
````

````python
# 4. Неоплаченные заказы: паттерн «накопитель»
unpaid_count = 0
unpaid_sum = 0
for a, p in zip(amounts_int, paid):
    if not p:
        unpaid_count += 1
        unpaid_sum += a

print(f"Не оплачено заказов: {unpaid_count}, на сумму {unpaid_sum:,} ₸")
````

````python
# 5. Таблица город | сумма | статус
print(f"{'город':<10} | {'сумма':>6} | статус")
print("-" * 32)
for c, a, p in zip(cities_clean, amounts_int, paid):
    status = "оплачен" if p else "НЕ оплачен"
    print(f"{c:<10} | {a:>6} | {status}")
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# 6. Выручка (только оплаченные) по Алматы и Астане отдельно
revenue_almaty = 0
revenue_astana = 0
for c, a, p in zip(cities_clean, amounts_int, paid):
    if p:
        if c == "Almaty":
            revenue_almaty += a
        elif c == "Astana":
            revenue_astana += a

print("Almaty:", revenue_almaty, "₸")
print("Astana:", revenue_astana, "₸")
````

````python
# 7. Самый крупный ОПЛАЧЕННЫЙ заказ (без max): запоминаем рекордсмена по ходу цикла
best_index = None
best_amount = 0
for i, (a, p) in enumerate(zip(amounts_int, paid)):
    if p and a > best_amount:
        best_amount = a
        best_index = i

print(f"Индекс: {best_index} (по счёту — заказ №{best_index + 1}), город: {cities_clean[best_index]}, сумма: {best_amount} ₸")
````

````python
# 8. Доля заказов выходного дня
weekend_count = 0
for d in days:
    if d == "сб" or d == "вс":
        weekend_count += 1

weekend_share = weekend_count / len(days) * 100
print(f"Заказов в выходные: {weekend_count} из {len(days)} = {weekend_share:.1f}%")
````

````python
# 9a. Копилка: каждый месяц одна и та же сумма
goal = 400000
monthly = 15000
saved = 0
months = 0
while saved < goal:
    saved += monthly
    months += 1

print(f"Откладывая {monthly:,} ₸ в месяц, накопим за {months} мес. (накоплено {saved:,} ₸)")
````

````python
# 9b. Копилка: каждый месяц откладываем на 1 000 ₸ больше, чем в прошлый
goal = 400000
deposit = 15000
saved = 0
months = 0
while saved < goal:
    saved += deposit
    months += 1
    deposit += 1000

print(f"С ежемесячной прибавкой 1 000 ₸ накопим за {months} мес. (накоплено {saved:,} ₸)")
````

## [Челлендж]
10. Без словарей и библиотек постройте два списка `unique_cities` и `revenue_by_city` (оплаченная выручка по каждому городу), отсортированные по убыванию выручки.
11. Найдите **медиану** суммы заказа без `statistics` и `numpy`. Учтите чётное число элементов.
12. Выведите текстовую «гистограмму» выручки по дням: каждые 5 000 ₸ рисуются как `█`.
```
пн | ████  20490
вт | ...
```
13. *Подумать:* какие проблемы данных вы встретили (пробелы, регистр, числа-строки)? Опишите 2–3 предложениями в Markdown-ячейке. Именно это мы будем делать в Pandas на занятии 6.

````python
# 10. Уникальные города и выручка по каждому — без словарей и библиотек
unique_cities = []
for c in cities_clean:
    if c not in unique_cities:
        unique_cities.append(c)

# revenue_by_city[i] — выручка города unique_cities[i] (та же позиция в обоих списках)
revenue_by_city = [0] * len(unique_cities)
for c, a, p in zip(cities_clean, amounts_int, paid):
    if p:
        i = unique_cities.index(c)
        revenue_by_city[i] += a

# Сортировка по убыванию выручки (пузырёк): меняем местами пары в ОБОИХ списках сразу
n = len(unique_cities)
for i in range(n):
    for j in range(n - 1 - i):
        if revenue_by_city[j] < revenue_by_city[j + 1]:
            revenue_by_city[j], revenue_by_city[j + 1] = revenue_by_city[j + 1], revenue_by_city[j]
            unique_cities[j], unique_cities[j + 1] = unique_cities[j + 1], unique_cities[j]

print(unique_cities)
print(revenue_by_city)
for c, r in zip(unique_cities, revenue_by_city):
    print(f"{c:<10} {r:>6} ₸")
````

````python
# 11. Медиана вручную (учитываем чётное количество элементов)
sorted_amounts = amounts_int.copy()      # копия, чтобы не испортить исходный список
sorted_amounts.sort()

n = len(sorted_amounts)
mid = n // 2
if n % 2 == 0:
    median = (sorted_amounts[mid - 1] + sorted_amounts[mid]) / 2
else:
    median = sorted_amounts[mid]

print("Отсортированные суммы:", sorted_amounts)
print("Медиана:", median, "₸")
````

````python
# 12. Гистограмма выручки (оплаченные заказы) по дням: 1 █ = 5 000 ₸ (неполные блоки отбрасываем)
day_order = ["пн", "вт", "ср", "чт", "пт", "сб", "вс"]

for day in day_order:
    day_revenue = 0
    for d, a, p in zip(days, amounts_int, paid):
        if d == day and p:
            day_revenue += a
    bar = "█" * (day_revenue // 5000)
    print(f"{day} | {bar:<9} {day_revenue}")
````

**13. Проблемы данных, которые я встретил.**
Города были записаны по-разному: с лишними пробелами (`" almaty"`, `"Almaty "`) и в разном регистре (`"SHYMKENT"`, `"astana"`), поэтому без очистки Python считал бы `"Astana"` и `"astana"` двумя разными городами. Суммы лежали строками (`"15990"`), и их нельзя было складывать, пока не преобразованы в `int`. Данные разнесены по четырём отдельным спискам, связанным только общим индексом, поэтому легко ошибиться и потерять соответствие, а считать по группам (например, выручку по городам) без словарей неудобно: приходится вручную вести два параллельных списка. Всё это в Pandas делается готовыми методами (`str.strip`, `astype`, `groupby`).
